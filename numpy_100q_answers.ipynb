````python
# **Q100.**  
# **Full ML pipeline challenge** — implement everything from scratch using only Sections 1–13:

# ```python
# X = np.random.rand(200, 5)
# y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)  # binary target
# ```

# 1. **Z-score standardize** `X` column-wise using broadcasting
# 2. **Shuffle and split** 70/15/15 using `np.argsort(np.random.rand(200))` as shuffle index
# 3. **Initialize weights** `w = np.zeros(5)` and bias `b = 0.0`
# 4. **Forward pass** on `X_train`: `z = X_train @ w + b`, `y_hat = 1 / (1 + np.exp(-z))`
# 5. **BCE loss**: clip `y_hat` to `[1e-7, 1-1e-7]`, compute `-np.mean(y*log + (1-y)*log)`
# 6. **Gradient**: `grad_w = X_train.T @ (y_hat - y_train) / len(y_train)`
# 7. **One gradient descent step**: `w -= 0.1 * grad_w`
# 8. **Forward pass again** with updated `w`, compute new BCE loss
# 9. Print: initial loss, updated loss (should be slightly lower), gradient norm `np.sqrt((grad_w**2).sum())`
````

````python

import numpy as np

np.random.seed(42)

# Generate dataset
X = np.random.rand(200, 5)
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)

# 1. Z-score standardization (column-wise)
mean = X.mean(axis=0)
std = X.std(axis=0)

# Avoid division by zero
std[std == 0] = 1

X_scaled = (X - mean) / std

# 2. Shuffle and split: 70% / 15% / 15%
indices = np.argsort(np.random.rand(200))

X_shuffled = X_scaled[indices]
y_shuffled = y[indices]

X_train = X_shuffled[:140]
y_train = y_shuffled[:140]

X_val = X_shuffled[140:170]
y_val = y_shuffled[140:170]

X_test = X_shuffled[170:]
y_test = y_shuffled[170:]

print("Train shape:", X_train.shape)
print("Validation shape:", X_val.shape)
print("Test shape:", X_test.shape)

# 3. Initialize weights and bias
w = np.zeros(5)
b = 0.0
learning_rate = 0.1

# Helper function: sigmoid
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Helper function: binary cross-entropy loss
def bce_loss(y_true, y_pred):
    y_pred = np.clip(y_pred, 1e-7, 1 - 1e-7)

    return -np.mean(
        y_true * np.log(y_pred)
        + (1 - y_true) * np.log(1 - y_pred)
    )

# 4. Initial forward pass
z = X_train @ w + b
y_hat = sigmoid(z)

# 5. Initial BCE loss
initial_loss = bce_loss(y_train, y_hat)

# 6. Calculate gradients
grad_w = X_train.T @ (y_hat - y_train) / len(y_train)
grad_b = np.mean(y_hat - y_train)

# Gradient norm (before updating weights)
gradient_norm = np.sqrt((grad_w ** 2).sum())

# 7. One gradient descent step
w -= learning_rate * grad_w
b -= learning_rate * grad_b

# 8. Forward pass with updated parameters
z_updated = X_train @ w + b
y_hat_updated = sigmoid(z_updated)

updated_loss = bce_loss(y_train, y_hat_updated)

# 9. Print results
print("\nInitial loss:", initial_loss)
print("Updated loss:", updated_loss)
print("Gradient norm:", gradient_norm)

print("\nUpdated weights:", w)
print("Updated bias:", b)

print("\nLoss decreased:", updated_loss < initial_loss)
````
